```python
RAW_FOLDER = r"C:\Users\HP\ISRO_Project\Minimum_Temperature\Raw_Data"
OUTPUT_FOLDER = r"C:\Users\HP\ISRO_Project\Minimum_Temperature\CSV_Data"
```

```python
latitudes = np.arange(7.5, 38.5, 1.0)
longitudes = np.arange(67.5, 98.5, 1.0)
```

```python
for year in range(2010, 2026):

    print(f"\n===== Processing {year} =====")

    file_path = os.path.join(
        RAW_FOLDER,
        f"Mintemp_MinT_{year}.GRD"
    )
    min_data = np.fromfile(file_path, dtype=np.float32)

    # automatic days ke liye 
    days = min_data.size // (31 * 31)

    # Reshape
    min_temp = min_data.reshape(days, 31, 31)

    # missing values 
    min_temp[min_temp == 99.9] = np.nan

    # dates
    dates = pd.date_range(
        start=f"{year}-01-01",
        periods=days,
        freq="D"
    )

    # records ke liye 
    records = []

    for day in range(days):
        for i, lat in enumerate(latitudes):
            for j, lon in enumerate(longitudes):
                records.append([
                    dates[day],
                    lat,
                    lon,
                    min_temp[day, i, j]
                ])

    # DataFrame ke liye 
    df_min = pd.DataFrame(
        records,
        columns=[
            "Date",
            "Latitude",
            "Longitude",
            "Min_Temperature"
        ]
    )

    # Save
    save_path = os.path.join(
        OUTPUT_FOLDER,
        f"df_min_{year}.csv"
    )

    df_min.to_csv(save_path, index=False)

    print(f"Saved: df_min_{year}.csv")
    print(f"Shape: {df_min.shape}")

print("\n All Minimum Temperature files processed successfully!")
```
