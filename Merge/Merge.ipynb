All quality chcekup down and Now I will run loop for all data set to merge 

```python
import os
MAX_FOLDER = r"C:\Users\HP\ISRO_Project\Max_Temperature\CSV_Data"
MIN_FOLDER = r"C:\Users\HP\ISRO_Project\Minimum_Temperature\CSV_Data"
RAIN_FOLDER = r"C:\Users\HP\ISRO_Project\Rainfall\CSV_Data"
OUTPUT_FOLDER = r"C:\Users\HP\ISRO_Project\Merge\Merged_CSV"

os.makedirs(OUTPUT_FOLDER, exist_ok=True)

for year in range(2010, 2026):

    print(f"\n===== Processing {year} =====")

    df_max = pd.read_csv(
        os.path.join(MAX_FOLDER, f"df_max_{year}.csv")
    )

    df_min = pd.read_csv(
        os.path.join(MIN_FOLDER, f"df_min_{year}.csv")
    )

    df_rain = pd.read_csv(
        os.path.join(RAIN_FOLDER, f"df_rain_{year}.csv")
    )

    # Max + Min
    df_merge = pd.merge(
        df_max,
        df_min,
        on=["Date", "Latitude", "Longitude"],
        how="inner"
    )
    # Rainfall
    df_final = pd.merge(
        df_merge,
        df_rain,
        on=["Date", "Latitude", "Longitude"],
        how="inner"
    )

    # missing rows
    df_clean = df_final.dropna()

    # Save
    output_path = os.path.join(
        OUTPUT_FOLDER,
        f"climate_data_{year}.csv"
    )

    df_clean.to_csv(output_path, index=False)

    print(f"Saved: climate_data_{year}.csv")
    print(f"Rows after cleaning: {df_clean.shape[0]}")

print("\nAll years merged successfully!")
```

```python
MERGED_FOLDER = r"C:\Users\HP\ISRO_Project\Merge\Merged_CSV"
print("=" * 70)
print("Checking all merged climate files...")
print("=" * 70)
for year in range(2010, 2026):
    file_path = os.path.join(MERGED_FOLDER, f"climate_data_{year}.csv")
    df = pd.read_csv(file_path)
    print(
        f"{year} | Shape: {df.shape} | "
        f"Missing: {df.isna().sum().sum()} | "
        f"Duplicates: {df.duplicated().sum()}"
    )
print("\nAll files checked successfully!")
```

```python
MERGED_FOLDER = r"C:\Users\HP\ISRO_Project\Merge\Merged_CSV"
FINAL_FOLDER = r"C:\Users\HP\ISRO_Project\Merge\Final_Dataset"
os.makedirs(FINAL_FOLDER, exist_ok=True)
all_dfs = []

print("=" * 60)
print("Combining climate data from 2010 to 2025...")
print("=" * 60)

# 2010 se 2025 tak sab files read karne ke liye hai 
for year in range(2010, 2026):
    file_path = os.path.join(
        MERGED_FOLDER,
        f"climate_data_{year}.csv"
    )

    df = pd.read_csv(file_path)
    all_dfs.append(df)

    print(f"✅ Loaded: climate_data_{year}.csv | Shape: {df.shape}")

# Sabhi DataFrames ko combine karne ke liye 
final_df = pd.concat(all_dfs, ignore_index=True)
print("\n" + "=" * 60)
print("Final Dataset Created Successfully!")
print("=" * 60)
print("Final Shape:", final_df.shape)
# Final CSV save karo
output_path = os.path.join(
    FINAL_FOLDER,
    "climate_data_2010_2025.csv"
)
final_df.to_csv(output_path, index=False)

print(f"\nSaved Successfully at:\n{output_path}")
```
