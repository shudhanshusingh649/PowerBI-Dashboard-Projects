```python
import numpy as np

data = np.fromfile("Rainfall_ind2024_rfp25.grd", dtype=np.float32)
rainfall = data.reshape(366, 129, 135)
rainfall[rainfall == -999] = np.nan

print(rainfall.shape)
print(rainfall[0, 0, 0])
print(rainfall[0, 50, 50])
```

```python
latitudes = np.arange(6.5, 6.5 + 129 * 0.25, 0.25)
longitudes = np.arange(66.5, 66.5 + 135 * 0.25, 0.25)

print(latitudes[:5])
print(longitudes[:5])
```

```python
print("Latitude:", latitudes[50])
print("Longitude:", longitudes[50])
print("Rainfall on 1 Jan 2024:", rainfall[0, 50, 50])
```

```python
import pandas as pd

records = []

for i, lat in enumerate(latitudes):
    for j, lon in enumerate(longitudes):
        records.append({
            "Date": "2024-01-01",
            "Latitude": lat,
            "Longitude": lon,
            "Rainfall": rainfall[0, i, j]
        })

df = pd.DataFrame(records)

print(df.head())
print(df.shape)
```

```python
print(df["Latitude"].min())
print(df["Latitude"].max())

print(df["Longitude"].min())
print(df["Longitude"].max())
```

```python
print("Latitude Range:", df["Latitude"].min(), "to", df["Latitude"].max())
print("Longitude Range:", df["Longitude"].min(), "to", df["Longitude"].max())
```

```python
df_rain_1deg = df.copy()

df_rain_1deg["Latitude"] = np.floor(df_rain_1deg["Latitude"]) + 0.5
df_rain_1deg["Longitude"] = np.floor(df_rain_1deg["Longitude"]) + 0.5
df_rain_1deg = (
    df_rain_1deg
    .groupby(["Date", "Latitude", "Longitude"], as_index=False)
    .agg({"Rainfall": "mean"})
)
df_rain_1deg = df_rain_1deg[
    (df_rain_1deg["Latitude"] >= 7.5) &
    (df_rain_1deg["Latitude"] <= 37.5) &
    (df_rain_1deg["Longitude"] >= 67.5) &
    (df_rain_1deg["Longitude"] <= 97.5)
]

print(df_rain_1deg.head())
print(df_rain_1deg.shape)
```

```python
print(df_rain_1deg["Latitude"].unique()[:5])
print(df_rain_1deg["Longitude"].unique()[:5])
```
