```python
import numpy as np
max_data = np.fromfile(
    "Maxtemp_MaxT_2024.GRD",  
    dtype=np.float32
)
print(max_data.shape)
```

```python
import numpy as np

# 99.9 ke bahut close values ko bhi NaN bana do
max_temp[np.isclose(max_temp, 99.9, atol=1e-3)] = np.nan
```

```python
max_temp = max_data.reshape(366, 31, 31)
print(max_temp.shape)
```

```python
max_temp[max_temp == 99.9] = np.nan
```

```python
latitudes_max = np.arange(7.5, 7.5 + 31 * 1, 1)
longitudes_max = np.arange(67.5, 67.5 + 31 * 1, 1)
print(latitudes_max)
print(longitudes_max)
```

```python
print(max_temp.shape)
print(max_temp[0, 0, 0])
print(max_temp[0, 10, 10])
```

```python
import pandas as pd
records = []
for i, lat in enumerate(latitudes_max):
    for j, lon in enumerate(longitudes_max):
        records.append({
            "Date": "2024-01-01",
            "Latitude": lat,
            "Longitude": lon,
            "Max_Temperature": max_temp[0, i, j]
        })
df_max = pd.DataFrame(records)
print(df_max.head())
print(df_max.shape)
```
