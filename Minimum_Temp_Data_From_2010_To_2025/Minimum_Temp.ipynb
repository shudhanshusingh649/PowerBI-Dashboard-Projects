```python
import numpy as np
min_data = np.fromfile(
    "Mintemp_MinT_2024.GRD",  
    dtype=np.float32
)
print(min_data.shape)
```

```python
min_temp = min_data.reshape(366, 31, 31)
```

```python
min_temp[min_temp == 99.9] = np.nan
```

```python
print(min_temp.shape)
print(min_temp[0, 0, 0])
print(min_temp[0, 10, 10])
```

```python
import pandas as pd
latitudes_min = np.arange(7.5, 7.5 + 31, 1)
longitudes_min = np.arange(67.5, 67.5 + 31, 1)
records = []
# Sirf 1 January 2024 ke liye
for i, lat in enumerate(latitudes_min):
    for j, lon in enumerate(longitudes_min):
        records.append({
            "Date": "2024-01-01",
            "Latitude": lat,
            "Longitude": lon,
            "Min_Temperature": min_temp[0, i, j]
        })
df_min = pd.DataFrame(records)
print(df_min.head())
print(df_min.shape)
```

```python
df_min.to_csv("df_min_2024.csv", index=False)
print("df_min saved successfully!")
```
