```python
RAW_FOLDER = r"C:\Users\HP\ISRO_Project\Max_Temperature\Raw_Data"
OUTPUT_FOLDER = r"C:\Users\HP\ISRO_Project\Max_Temperature\CSV_Data"
```

```python
latitudes = np.arange(7.5, 38.5, 1.0)
longitudes = np.arange(67.5, 98.5, 1.0)
```

```python
for year in range(2010, 2026):

    print(f"\n===== Processing {year} =====")
    
    file_path = os.path.join(
        RAW_FOLDER,
        f"Maxtemp_MaxT_{year}.GRD"
    )

    # Read binary data
    max_data = np.fromfile(file_path, dtype=np.float32)

    # Number of days automatically 
    days = max_data.size // (31 * 31)

    # Reshape
    max_temp = max_data.reshape(days, 31, 31)

    # missing value
    max_temp[max_temp == 99.9] = np.nan

    # For dates
    dates = pd.date_range(
        start=f"{year}-01-01",
        periods=days,
        freq="D"
    )

    # records
    records = []

    for day in range(days):
        for i, lat in enumerate(latitudes):
            for j, lon in enumerate(longitudes):
                records.append([
                    dates[day],
                    lat,
                    lon,
                    max_temp[day, i, j]
                ])

    # DataFrame
    df_max = pd.DataFrame(
        records,
        columns=[
            "Date",
            "Latitude",
            "Longitude",
            "Max_Temperature"
        ]
    )

    # Save
    save_path = os.path.join(
        OUTPUT_FOLDER,
        f"df_max_{year}.csv"
    )

    df_max.to_csv(save_path, index=False)

    print(f"Saved: df_max_{year}.csv")
    print(f"Shape: {df_max.shape}")
```
