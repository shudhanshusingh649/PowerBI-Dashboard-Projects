```python
import pandas as pd
df_max = pd.read_csv(
    r"C:\Users\HP\ISRO_Project\Maxmium_Temp_Data_from_2010_To_2025\df_max_2024.csv")
df_min = pd.read_csv(
    r"C:\Users\HP\ISRO_Project\Minimum_Temp_Data_From_2010_To_2025\df_min_2024.csv")
df_rain = pd.read_csv(
    r"C:\Users\HP\ISRO_Project\Rainfall_Data_From_2010_To_2025\df_rain_1deg.csv")
df_temp = pd.merge(
    df_max,
    df_min,
    on=["Date", "Latitude", "Longitude"],
    how="inner")
df_final = pd.merge(
    df_temp,
    df_rain,
    on=["Date", "Latitude", "Longitude"],
    how="inner"
)
```

```python
df_clean = df_final.dropna()
print(df_clean.head())
print(df_clean.shape)
```

```python
df_clean.to_csv(
    r"C:\Users\HP\ISRO_Project\Merge_file\climate_data_2024.csv",
    index=False
)

print("Climate dataset saved successfully!")
```
